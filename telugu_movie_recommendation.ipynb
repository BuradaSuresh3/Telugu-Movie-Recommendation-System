```python
# Step 1: Import Required Libraries

import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import streamlit as st


# Step 2: Load Telugu Movie Dataset

movies = pd.read_csv("dataset/telugu_movies_2021_2026_50_plus.csv")

print("Dataset Preview:")
print(movies.head())

print("\nColumn Names:")
print(movies.columns)

print("\nDataset Shape:")
print(movies.shape)

print("\nMissing Values:")
print(movies.isnull().sum())


# Step 3: Create Combined Features

movies["combined_features"] = (
    movies["Genre/Type"] + " " +
    movies["Hero"] + " " +
    movies["Director"]
)

print("\nUpdated Dataset:")
print(
    movies[
        ["Movie Name", "combined_features"]
    ].head()
)


# Step 4: Convert Text Data into Numerical Matrix

cv = CountVectorizer()

count_matrix = cv.fit_transform(
    movies["combined_features"]
)

print("\nCount Matrix Shape:")
print(count_matrix.shape)


# Step 5: Calculate Cosine Similarity

similarity = cosine_similarity(
    count_matrix
)

print("\nSimilarity Matrix Shape:")
print(similarity.shape)


# Step 6: Create Movie Index Mapping

indices = pd.Series(
    movies.index,
    index=movies["Movie Name"]
).drop_duplicates()


# Step 7: Recommendation Function

def recommend_movies(movie_name):
    idx = indices[movie_name]

    similarity_scores = list(
        enumerate(similarity[idx])
    )

    similarity_scores = sorted(
        similarity_scores,
        key=lambda x: x[1],
        reverse=True
    )

    # Skip same movie
    similarity_scores = similarity_scores[1:6]

    recommended_movies = []

    for i in similarity_scores:
        recommended_movies.append(
            movies.iloc[i[0]]["Movie Name"]
        )

    return recommended_movies


# Step 8: Test Recommendation System

print("\nTop Recommendations for Pushpa: The Rise\n")

recommendations = recommend_movies(
    "Pushpa: The Rise"
)

for movie in recommendations:
    print(movie)


# Step 9: Streamlit UI

st.title("🎬 Telugu Movie Recommendation System")

st.write(
    "Select a Telugu movie and get similar movie recommendations!"
)

selected_movie = st.selectbox(
    "Select a Movie",
    movies["Movie Name"].values
)

if st.button("Recommend"):
    recommendations = recommend_movies(
        selected_movie
    )

    st.subheader(
        f"Top Recommendations for: {selected_movie}"
    )

    for movie in recommendations:
        movie_details = movies[
            movies["Movie Name"] == movie
        ].iloc[0]

        st.write(
            f"🎥 {movie_details['Movie Name']} | "
            f"⭐ Rating: {movie_details['Rating']} | "
            f"🎭 Genre: {movie_details['Genre/Type']} | "
            f"👤 Hero: {movie_details['Hero']} | "
            f"🎬 Director: {movie_details['Director']}"
        )
```
